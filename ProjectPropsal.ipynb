## Dataset Description:
[SupplyChainGHGEFactors](https://catalog.data.gov/dataset/supply-chain-greenhouse-gas-emission-factors-v1-2-by-naics-6?from_hint=eyJzb3J0IjoicG9wdWxhcml0eSJ9)

The data from the site above is imported the form of a CSV file containing 8 columns of data and just over 1000 rows. 

2017 NAICS Code
: This is the North American Industry Classification System (NAICS) identifier number. The number is just a representation in the system. All the values in this list are 6 digit identifiers excluding households, electricity, and government commodities.

2017 NAICS Title
: The title of the object/commodity that hte NAICS code refers too.

GHG
: Green House Gas (GHG) is in refrence to which type of greenhouse gas that the object/commodity produces.

Unit
: The metric in which the amount of GHG emmitted is measured in.

Supply Chain Emission Factors without Margins (SCEF)
: The amount of GHG produced by the creation/action of the commodity excluding the transportation or distribution.

Margins of Supply Chain Emission Factors (MSCEF)
: The amoutn of GHG produced by the transportation of the commodity

Supply Chain Emission Factors with Margins (SCEFM)
: The combination of the previous two emission factors.

Reference USEEIO Code
: US Environmentally-Extended Input-Output (USEEIO) refrence code is a code that ties directly to a sector/product coming from the USEEIO models/database.

## Classifier and Prediction Plan
The data set contains three sets of values that are worth predicting, SCEF, MSCEF, SCEFM. The main value worth tracking is the SCEFM which is the GHG production with the margins of transportation and handeling included. The value for tracking and predicting this is that it gives the most accurate prediction to the total GHG production of a certain commodity. We plan to use three or four classifiers during our data handeling. The first three are Decision Tree, Random Forest, and K-NN. If time allows or it seems valuable to the predicition of the SCEFM value we would use Naive Bayes as well. The end goal is to have a model that can accurately predict the SCEFM for future use based on previous years produced values.

The final predictions done on the SCEFM will need be generalized into the three classes that we are using. The generalized classes are good, acceptable, and bad. The ranges for good, acceptable, and bad are as follows: 0.01 - 0.185 (good), 0.186 - 0.386 (acceptable), 0.387 - 10.99 (bad). The goal of using these three classes is to give a predictable estimate to the continuous values that the data set provides us. These estimates can be used by people in the field to make decisions for the coming years and used as comparisons based on their own previously collected data.
